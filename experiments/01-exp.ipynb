```python
import os
from getpass import getpass
from dotenv import load_dotenv
load_dotenv()
```

```python
from meshapi import MeshAPI
MESHAPI_TOKEN = os.getenv("MESH_API_KEY")

client = MeshAPI(
    base_url=os.getenv("MESHAPI_BASE_URL"),
    token=MESHAPI_TOKEN
)

print("client ready")
```

```python
models = client.models.list()

free_models = [
    model for model in models
    if model.is_free
]

free_models
```

```python
from meshapi import ChatCompletionParams, ChatMessage
resp = client.chat.completions.create(
    ChatCompletionParams(
        model="minimax/m2-her",
        messages=[ChatMessage(
            role="user",
            content="Give me a one line explaination of AI Gateways"
        )],
        max_tokens=60
    )
)

resp.choices[0].message.content
```
